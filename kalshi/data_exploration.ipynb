```python
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
```

```python
from kalshi.events import Events

events = Events()
events = events.get_all_events(all_pages=True, status='open')
events_df = pd.DataFrame(events)
display(events_df.head())
```

```python
display(events_df.columns)
```

```python
display(events_df.describe().transpose())
```

```python
display(events_df['category'].value_counts())
```

```python
display(events_df[events_df['category'] == 'Sports'])
```

```python
from kalshi.markets import Markets
from kalshi.utils import utc_to_local
markets = Markets()
list_markets = markets.get_all_markets(all_pages=False, status='open', mve_filter='exclude')
markets_df = pd.DataFrame(list_markets)
cols = markets_df.columns
for col in cols:
    if 'time' in col:
        markets_df[col] = markets_df[col].apply(utc_to_local)
```

```python
display(markets_df.sort_values('close_time', ascending=True).head())
```

```python
display(markets_df)
```

```python
display(markets_df.describe().transpose())
```
